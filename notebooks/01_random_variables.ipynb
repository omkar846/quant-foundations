```python
import numpy as np
```

```python
rng = np.random.default_rng(42)
```

```python
rolls = rng.integers(1, 7, size=100000)
```

```python
print("Mean:", np.mean(rolls))
print("Variance:", np.var(rolls))
print("Standard Deviation:", np.std(rolls))
```

```python
# Frequency of each dice face
for face in range(1, 7):
    count = np.sum(rolls == face)
    print(face, count, count / len(rolls))

# How the sample mean changes with sample size
for n in [10, 100, 1000, 10000, 100000]:
    print(n, np.mean(rolls[:n]))
```
